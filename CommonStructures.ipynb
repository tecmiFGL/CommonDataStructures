#Variables
Variables are containers of information that allow us to use data anywhere in the code, we can also modify the value

##Data types
Variables can contain any kind of information or data; there are different types of data commonly used in python, these are:

###Integer     | int()
    It stores whole/integer numbers
    We declare an int variable the following ways:
        variableName = int()
        variableName = 3    # or any integer number

###Float       | float()
    It stores decimals (e.g. 3.1416)
    We declare a float variable the following ways:
        variableName = float()
        variableName = 3.1416    # or any decimal number

###String      | str()
    It stores letters, words or phrases. Strings in python always go in between quotation marks ("")
    We declare a string variable the following ways:
        variableName = str()
        variableName = "This is an string variable"    # or letter, word or phrase

###Boolean     | bool()
    It stores True or False
    We declare a boolean variable the following ways:
        variableName = True
        variableName = False
        fun fact:   
    An empty variable [eg. emptyVariable = int()] is False
    A filled variable [eg. notEmptyVariable = 4 ] is True

###List        | list()
    A list is a variable that can contain multiple elements. These elements are stored in a position, thing of this like a tray of chocolates, where each chocolate has its own place
    We declare a list variable the following ways:
        variableName = list()
        variableName = [3, "name", 56.3, True]    # we can store any type of data inside a list, and as many elements as we want

```python
variableIntType = int()     #empty variable
variableIntType = 5
variableIntType = int(input(""))    #ask user to input an integer number

variableFloatType = float() #empty variable
variableFloatType = 3.1416
variableFloatType = float(input(""))    #ask user to input a decimal

variableSrtingType = str()
variableStringType = "Name"
variableStringType = input("")    #ask user to input anything, when using input alone even a number becomes a [string]

variableBoolType = bool()   #empty Variable
variableBoolType = True #for true
variableBoolType = False #for false
variableBoolType = bool(input(""))    #ask user to input anything, if no input detected = False, if something is input = True

variableListType = list()   # create empty list
variableListType = []       # create empty list
variableListType = [3, "name", 56.3, True]
variableListType = [1,2,3,4]
variableListType = ["apple", "banana", "grape"]
variableListType = list(input(""))    #any user input will be segmented into elements by character, everything is a string , eg. input: hello12, list=['h', 'e', 'l', 'l', 'o', '1', '2']
```

#For loop
A for loop is a cycle that allows a block of code (anything inside the for) to be repeated as many times as stated

Syntax
###When we use range(int()) we are telling the loop to repeat n number of times
    for controlVariable in range(3):
        #   block of code that will repeat 3 times...
        #   at the beginning, first iteration controlVariable = 0
        #   in the second iteration controlVariable = 1
        #   in the third iteration controlVariable = 2

###When we use range(int(),int()) we are telling the loop to repeat starting from the first int() until the second int(). This means that the controlVariable will add one (starting from the first int()) each time it completes the block of code, and until it reaches the second int()-1
    for controlVariable in range(1,5):
        #   block of code that will repeat...
        #   first iteration     controlVariable = 1
        #   second iteration    controlVariable = 2
        #   third iteration     controlVariable = 3
        #   the last iteration  controlVariable = 4

###When we use range(int(),int(),int()) we are telling the loop to repeat starting from the first int() until the second int(), and the third int() tells how much we will add each time the loop completes the block of code one time (starting from the first int()) each time it completes the block of code, and until the controlVariable < int()
    for controlVariable in range(1,5,2):
        #   block of code that will repeat...
        #   first iteration     controlVariable = 1
        #   last iteration      controlVariable = 3


```python
print("for loop syntax (simple)")
for controlVariable in range(3):
    print(controlVariable)
print("\n for loop using start and stop" )
for controlVariable in range(1,5):
    print(controlVariable)
print("\n for loop using start stop and step" )
for controlVariable in range (1,10,2):
    print(controlVariable)
```

#While loop
A while loop is a cycle that allows a block of code (anything inside the while) to be repeated as long as a condition is evaluated True

##Syntax
while condition:    #condition == True
    print("Inside while")

Example above will go on and on until the end of life, thats why we have to change the value inside the while

while condition:
    print("inside while")
    condition=False #change the value condition so the loop stops after its first iteration

We can also use counters or any other comparison expression


```python
condition=True
while condition:
    print("Inside while")
    condition=False
print("Out of while")
```

#for loops and lists 
We can iterate (go over) a list of elements with a for loop, in this case the controlVariable takes the value of a position in a list. When using for with lists the element position starts in 0.

variableName = ["a,","b,","c","d"]
for controlVariable in variableName:
        print(variableName)
        #   block of code that will repeat...
        #   first iteration     controlVariable = "a"
        #   second iteration    controlVariable = "b"
        #   third iteration     controlVariable = "c"
        #   the last iteration  controlVariable = "d"

We can also go through the same list of elements with a for loop using range(int()) to point the position of the element in the list. We just need the number of elements in the list.
variableName = ["a,","b,","c","d"]
elements_in_list = 4
for controlVariable in range(elements_in_list):
        #   block of code that will repeat...
        print(variableName[controlVariable])
        #   first iteration     controlVariable = 0    "a"
        #   second iteration    controlVariable = 1    "b"
        #   third iteration     controlVariable = 2    "c"
        #   the last iteration  controlVariable = 3    "d"

```python
print("Iterating over lists")
variableName = ["a","b","c","d"]
for controlVariable in variableName:
    print(controlVariable)

print("Using lists")
variableName = ["a","b","c","d"]
elements_in_list = 4
elements_in_list = len(variableName) # we can also use the function len(variableName) to get the number of elements on a list
for controlVariable in range(elements_in_list):
        print(variableName[controlVariable])
```
